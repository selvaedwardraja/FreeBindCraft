```python
import json
from pathlib import Path

# Start from the upstream FreeBindCraft configuration
with open("settings_advanced/default_4stage_multimer.json") as f:
    advanced = json.load(f)

# Only modify parameters needed for the Colab smoke test
advanced["save_design_animations"] = False
advanced["save_design_trajectory_plots"] = False
advanced["max_trajectories"] = 1
advanced["enable_rejection_check"] = False

# Use the AF2 weights and DSSP installed in this Colab runtime
advanced["af_params_dir"] = "/content/FreeBindCraft/params"
advanced["dssp_path"] = "/content/FreeBindCraft/functions/dssp"

# PyRosetta is disabled for this Colab implementation
advanced["dalphaball_path"] = ""

smoke_advanced_path = "settings_advanced/colab_smoke_test.json"

with open(smoke_advanced_path, "w") as f:
    json.dump(advanced, f, indent=2)

print("Updated smoke-test advanced settings:")
print(smoke_advanced_path)

print("\nKey parameters:")
for key in [
    "use_termini_distance_loss",
    "save_design_animations",
    "save_design_trajectory_plots",
    "max_trajectories",
    "enable_rejection_check",
    "af_params_dir",
    "dssp_path"
]:
    print(f"{key}: {advanced.get(key)}")
```

```python
import pandas as pd

result_dir = "/content/FreeBindCraft/colab_smoke_test"
df = pd.read_csv(f"{result_dir}/final_design_stats.csv")

# Show columns containing important ranking/interface metrics
keywords = [
    "i_pTM",
    "ipSAE",
    "pTM",
    "pLDDT",
    "PAE",
    "RMSD",
    "interface",
    "contact",
    "binder"
]

selected = [
    c for c in df.columns
    if any(k.lower() in c.lower() for k in keywords)
]

print("Selected scoring/interface columns:")
print("\n".join(selected))

print("\n=== KEY RESULTS ===")
display(df[["Rank", "Design", "Length"] + selected])
```
