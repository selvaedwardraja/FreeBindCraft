```python
# ============================================================
# 🧬 FreeBindCraft — Cell 5: Configure Conda
# ============================================================

from pathlib import Path
import subprocess
import os

print("=" * 72)
print("🧬 FreeBindCraft — Step 5: Conda Configuration")
print("=" * 72)
print()
print("STATUS: 🔄 Preparing Conda for FreeBindCraft...")
print()

CONDA = "/usr/local/miniconda/bin/conda"

# ------------------------------------------------------------
# HELPER
# ------------------------------------------------------------
def run_conda(args, show_output=False):
    result = subprocess.run(
        [CONDA] + args,
        capture_output=True,
        text=True
    )

    if show_output:
        if result.stdout:
            print(result.stdout)
        if result.stderr:
            print(result.stderr)

    return result


# ------------------------------------------------------------
# STEP 1 — VERIFY CONDA
# ------------------------------------------------------------
print("-" * 72)
print("STEP 1/4 — Verifying Conda")
print("-" * 72)

result = run_conda(["--version"])

if result.returncode != 0:
    raise RuntimeError(
        "Conda could not be executed."
    )

print(f"✓ {result.stdout.strip()}")

# ------------------------------------------------------------
# STEP 2 — CONFIGURE CHANNELS
# ------------------------------------------------------------
print()
print("-" * 72)
print("STEP 2/4 — Configuring Conda channels")
print("-" * 72)

print("⏳ Configuring:")
print("   • defaults")
print("   • pkgs/main")
print("   • pkgs/r")
print()

# Ensure defaults is available
result = run_conda([
    "config",
    "--add",
    "channels",
    "defaults"
])

if result.returncode != 0:
    print("⚠️ Channel configuration returned a message:")
    print(result.stderr)

print("✓ Conda channel configuration updated.")

# ------------------------------------------------------------
# STEP 3 — ACCEPT ANACONDA TERMS
# ------------------------------------------------------------
print()
print("-" * 72)
print("STEP 3/4 — Checking Anaconda Terms of Service")
print("-" * 72)

print("⏳ Checking Terms of Service status...")
print()

# Accept TOS for the repositories used by the installer.
tos_channels = [
    "https://repo.anaconda.com/pkgs/main",
    "https://repo.anaconda.com/pkgs/r",
]

for channel in tos_channels:

    print(f"⏳ Checking: {channel}")

    tos_result = subprocess.run(
        [
            CONDA,
            "tos",
            "accept",
            "--override-channels",
            "--channel",
            channel
        ],
        capture_output=True,
        text=True
    )

    if tos_result.returncode == 0:
        print("✓ Terms accepted/already accepted.")
    else:
        # Some Conda versions may report that the terms were
        # already accepted or may provide informational output.
        combined = (
            (tos_result.stdout or "") +
            (tos_result.stderr or "")
        ).strip()

        if "already" in combined.lower() or \
           "accepted" in combined.lower():

            print("✓ Terms were already accepted.")

        else:

            print("⚠️ Conda returned:")
            print(combined)

            raise RuntimeError(
                "Unable to configure Anaconda Terms of Service."
            )

# ------------------------------------------------------------
# STEP 4 — FINAL CONFIGURATION CHECK
# ------------------------------------------------------------
print()
print("-" * 72)
print("STEP 4/4 — Verifying Conda configuration")
print("-" * 72)

result = run_conda([
    "config",
    "--show",
    "channels"
])

if result.returncode != 0:
    raise RuntimeError(
        "Unable to read Conda channel configuration."
    )

print("📚 Active Conda channels:")
print()

print(result.stdout.strip())

# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------
print()
print("=" * 72)
print("STATUS: ✅ CONDA READY")
print("=" * 72)
print()

print("✓ Conda is operational")
print("✓ Required channels configured")
print("✓ Anaconda Terms of Service accepted")
print("✓ Ready for FreeBindCraft environment installation")

print()
print("👉 USER ACTION")
print("   No additional button/click is required.")
print()
print("👉 NEXT STEP")
print("   Run Cell [6] to install the PyRosetta-free")
print("   FreeBindCraft environment.")
print()
print("⚠️ IMPORTANT")
print("   Cell [6] will perform the main software installation.")
print("   It may take several minutes and will download")
print("   the AlphaFold model parameters.")
```

```python
# ============================================================
# 🧬 FreeBindCraft — Cell 6: Install FreeBindCraft Environment
# Compact progress version
# ============================================================

from pathlib import Path
import subprocess
import shutil
import os
import time
import sys
import re
import threading
from IPython.display import display, HTML, clear_output

# ============================================================
# CONFIGURATION
# ============================================================

REPO_DIR = Path("/content/FreeBindCraft")
CONDA = "/usr/local/miniconda/bin/conda"
INSTALL_LOG = REPO_DIR / "freebindcraft_install.log"

BINDCRAFT_ENV = Path(
    "/usr/local/miniconda/envs/BindCraft"
)

# ============================================================
# UI HELPERS
# ============================================================

def progress_bar(percent, label="", width=520):
    percent = max(0, min(100, int(percent)))

    return f"""
    <div style="
        width:{width}px;
        max-width:100%;
        background:#252525;
        border:1px solid #555;
        border-radius:8px;
        overflow:hidden;
        height:14px;
        margin-top:6px;
        margin-bottom:4px;">
        <div style="
            width:{percent}%;
            height:100%;
            background:#22c55e;">
        </div>
    </div>

    <div style="
        font-family:Arial,sans-serif;
        font-size:12px;
        color:#d1d5db;">
        {percent}% {label}
    </div>
    """

def status_panel(title, message, kind="info"):
    styles = {
        "info":    ("#111827", "#60a5fa", "#f8fafc"),
        "success": ("#052e16", "#22c55e", "#f0fdf4"),
        "warning": ("#422006", "#f59e0b", "#fff7ed"),
        "error":   ("#450a0a", "#ef4444", "#fef2f2"),
    }

    bg, border, text = styles[kind]

    return f"""
    <div style="
        background:{bg};
        border:1px solid {border};
        border-radius:10px;
        padding:14px 16px;
        margin:8px 0;
        color:{text};
        font-family:Arial,sans-serif;
        font-size:13px;
        line-height:1.5;">

        <div style="
            font-size:15px;
            font-weight:700;
            margin-bottom:6px;">
            {title}
        </div>

        <div>{message}</div>
    </div>
    """

# ============================================================
# HEADER
# ============================================================

clear_output(wait=True)

display(HTML(
    f"""
    <div style="
        font-family:Arial,sans-serif;
        color:#f8fafc;
        background:#0f172a;
        border-radius:10px;
        padding:14px 16px;
        margin-bottom:10px;">

        <div style="font-size:19px;font-weight:700;">
            🧬 FreeBindCraft — Step 6: Software Installation
        </div>

        <div style="
            margin-top:5px;
            font-size:13px;
            color:#cbd5e1;">
            PyRosetta-free installation with compact progress monitoring
        </div>
    </div>
    """
))

display(HTML(progress_bar(0, "Preparing installation...")))

# ============================================================
# STEP 1 — BASIC VALIDATION
# ============================================================

print("=" * 72)
print("STEP 1/6 — Pre-installation checks")
print("=" * 72)

if not REPO_DIR.exists():
    raise RuntimeError(
        "FreeBindCraft repository was not found."
    )

if not (REPO_DIR / "install_bindcraft.sh").exists():
    raise RuntimeError(
        "install_bindcraft.sh was not found."
    )

if not Path(CONDA).exists():
    raise RuntimeError(
        "Miniconda was not found at /usr/local/miniconda."
    )

print("✓ FreeBindCraft repository found")
print("✓ Installation script found")
print("✓ Conda executable found")

display(HTML(progress_bar(10, "Pre-installation checks complete")))

# ============================================================
# STEP 2 — GPU CHECK
# ============================================================

print()
print("=" * 72)
print("STEP 2/6 — GPU check")
print("=" * 72)

gpu_result = subprocess.run(
    [
        "nvidia-smi",
        "--query-gpu=name,memory.total,memory.free",
        "--format=csv,noheader"
    ],
    capture_output=True,
    text=True
)

if gpu_result.returncode != 0:
    raise RuntimeError(
        "NVIDIA GPU is not available."
    )

gpu_info = gpu_result.stdout.strip()

print("✓ NVIDIA GPU detected")
print(f"  {gpu_info}")

display(HTML(progress_bar(15, "GPU check complete")))

# ============================================================
# STEP 3 — STORAGE CHECK
# ============================================================

print()
print("=" * 72)
print("STEP 3/6 — Storage check")
print("=" * 72)

disk = shutil.disk_usage("/")

free_gb = disk.free / (1024 ** 3)

print(f"✓ Free storage: {free_gb:.1f} GB")

if free_gb < 15:
    raise RuntimeError(
        "Insufficient free storage for FreeBindCraft installation."
    )

print("✓ Storage appears sufficient")

display(HTML(progress_bar(20, "Storage check complete")))

# ============================================================
# STEP 4 — EXISTING ENVIRONMENT CHECK
# ============================================================

print()
print("=" * 72)
print("STEP 4/6 — Checking existing BindCraft environment")
print("=" * 72)

env_result = subprocess.run(
    [CONDA, "env", "list"],
    capture_output=True,
    text=True
)

if env_result.returncode != 0:
    raise RuntimeError(
        "Unable to query Conda environments."
    )

bindcraft_exists = any(
    line.strip().startswith("BindCraft ")
    or "/envs/BindCraft" in line
    for line in env_result.stdout.splitlines()
)

if bindcraft_exists:

    print("🟢 Existing BindCraft environment detected.")
    print("No reinstallation will be performed.")

    display(HTML(progress_bar(
        100,
        "Existing environment detected — installation skipped"
    )))

else:

    print("ℹ️ BindCraft environment is not installed.")
    print("✓ A new environment will be created.")

    # ========================================================
    # STEP 5 — INSTALLATION
    # ========================================================

    print()
    print("=" * 72)
    print("STEP 5/6 — Installing FreeBindCraft")
    print("=" * 72)

    print()
    print("⚙️ Installation configuration")
    print("   Package manager : conda")
    print("   CUDA target     : 12.4")
    print("   PyRosetta       : OFF")
    print("   GPU             : A100 40 GB")

    print()
    print("📦 Major components")
    print("   • Python 3.10 environment")
    print("   • JAX / JAXlib")
    print("   • CUDA dependencies")
    print("   • AlphaFold dependencies")
    print("   • ColabDesign")
    print("   • OpenMM / PDBFixer")
    print("   • AlphaFold model parameters")

    print()
    print("🔄 Installer output is being captured silently.")
    print(f"📄 Full installer log: {INSTALL_LOG}")

    # --------------------------------------------------------
    # Installation command
    # --------------------------------------------------------

    os.chdir(REPO_DIR)

    install_cmd = [
        "bash",
        "install_bindcraft.sh",
        "--no-pyrosetta",
        "--cuda",
        "12.4",
        "--pkg_manager",
        "conda"
    ]

    start_time = time.time()

    # --------------------------------------------------------
    # Open persistent installation log
    # --------------------------------------------------------

    log_file = open(
        INSTALL_LOG,
        "w",
        encoding="utf-8",
        errors="replace"
    )

    process = subprocess.Popen(
        install_cmd,
        cwd=str(REPO_DIR),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    # --------------------------------------------------------
    # Shared status
    # --------------------------------------------------------

    state = {
        "latest_line": "Starting installer...",
        "download_percent": None,
        "stage": "Preparing installation",
        "lines": 0,
    }

    # --------------------------------------------------------
    # Recognize useful installer stages
    # --------------------------------------------------------

    def classify_line(line):

        low = line.lower()

        if "downloading" in low:
            return "Downloading package"

        if "download" in low and "%" in low:
            return "Downloading"

        if "alphafold" in low and (
            "param" in low
            or "weight" in low
        ):
            return "Downloading AlphaFold parameters"

        if "creating" in low and "environment" in low:
            return "Creating Conda environment"

        if "environment" in low:
            return "Configuring environment"

        if "installing" in low:
            return "Installing dependencies"

        if "pip" in low:
            return "Installing Python packages"

        if "conda" in low:
            return "Installing Conda packages"

        if "jax" in low:
            return "Installing JAX/CUDA components"

        if "openmm" in low:
            return "Installing OpenMM"

        if "colabdesign" in low:
            return "Installing ColabDesign"

        if "pdbfixer" in low:
            return "Installing PDBFixer"

        if "cleanup" in low or "cleaning" in low:
            return "Cleaning installation"

        if "complete" in low or "success" in low:
            return "Finalizing installation"

        return None

    def extract_percent(line):

        # Detect conventional percentages such as:
        # 23%
        # 23.5%
        matches = re.findall(
            r'(?<!\d)(\d{1,3}(?:\.\d+)?)\s*%',
            line
        )

        if not matches:
            return None

        try:
            value = float(matches[-1])

            if 0 <= value <= 100:
                return value

        except Exception:
            pass

        return None

    # --------------------------------------------------------
    # Reader thread
    # --------------------------------------------------------

    def read_installer_output():

        for raw_line in process.stdout:

            line = raw_line.rstrip()

            if not line:
                continue

            state["lines"] += 1
            state["latest_line"] = line

            stage = classify_line(line)

            if stage:
                state["stage"] = stage

            pct = extract_percent(line)

            if pct is not None:
                state["download_percent"] = pct

            try:
                log_file.write(line + "\n")
                log_file.flush()
            except Exception:
                pass

        try:
            log_file.flush()
            log_file.close()
        except Exception:
            pass

    reader = threading.Thread(
        target=read_installer_output,
        daemon=True
    )

    reader.start()

    # --------------------------------------------------------
    # Compact progress display
    # --------------------------------------------------------

    # IMPORTANT:
    # This is NOT a fake claim of installation percentage.
    # It is a UI progress indicator based on installer activity.
    #
    # 20–85% = installation running
    # 100%    = installer exited successfully

    last_render = 0

    while process.poll() is None:

        now = time.time()

        if now - last_render >= 2:

            elapsed = int(now - start_time)

            # Activity-based display.
            # Keep it below 95% until the process actually exits.
            if state["download_percent"] is not None:

                # Download percentage is genuine installer-reported
                # progress, but keep overall installation bar capped.
                overall = min(
                    90,
                    25 + int(
                        state["download_percent"] * 0.65
                    )
                )

            else:

                # Installer stage progress indicator.
                overall = 25

                stage_low = state["stage"].lower()

                if "creating" in stage_low:
                    overall = 30
                elif "configuring" in stage_low:
                    overall = 40
                elif "conda" in stage_low:
                    overall = 50
                elif "python" in stage_low:
                    overall = 55
                elif "jax" in stage_low:
                    overall = 60
                elif "alphafold" in stage_low:
                    overall = 70
                elif "openmm" in stage_low:
                    overall = 75
                elif "colabdesign" in stage_low:
                    overall = 78
                elif "cleanup" in stage_low:
                    overall = 90
                elif "final" in stage_low:
                    overall = 92

            elapsed_text = (
                f"{elapsed // 60:02d}:"
                f"{elapsed % 60:02d}"
            )

            percent_text = ""

            if state["download_percent"] is not None:
                percent_text = (
                    f"<br><b>Installer-reported download:</b> "
                    f"{state['download_percent']:.1f}%"
                )

            clear_output(wait=True)

            display(HTML(
                status_panel(
                    "🔄 FreeBindCraft installation in progress",
                    f"""
                    <b>Current stage:</b> {state['stage']}<br>
                    <b>Elapsed:</b> {elapsed_text}<br>
                    <b>Installer messages captured:</b> {state['lines']}
                    {percent_text}<br><br>

                    <b>Installer output is being saved silently to:</b><br>
                    <code>{INSTALL_LOG}</code>
                    """,
                    "info"
                )
            ))

            display(HTML(
                progress_bar(
                    overall,
                    state["stage"]
                )
            ))

            last_render = now

        time.sleep(0.5)

    # --------------------------------------------------------
    # Wait for output reader
    # --------------------------------------------------------

    reader.join(timeout=10)

    return_code = process.wait()

    elapsed_minutes = (
        time.time() - start_time
    ) / 60

    # --------------------------------------------------------
    # Installation result
    # --------------------------------------------------------

    if return_code != 0:

        clear_output(wait=True)

        display(HTML(
            status_panel(
                "❌ INSTALLATION FAILED",
                f"""
                FreeBindCraft installation exited with code
                <b>{return_code}</b>.<br><br>

                The full installer output has been preserved at:<br>
                <code>{INSTALL_LOG}</code><br><br>

                No design has been started.
                """,
                "error"
            )
        ))

        raise RuntimeError(
            f"FreeBindCraft installation failed "
            f"with exit code {return_code}."
        )

    clear_output(wait=True)

    display(HTML(
        status_panel(
            "✅ INSTALLATION COMPLETED",
            f"""
            FreeBindCraft installation completed successfully.<br><br>

            <b>Elapsed time:</b> {elapsed_minutes:.1f} minutes<br>
            <b>PyRosetta:</b> OFF<br>
            <b>CUDA target:</b> 12.4<br>
            <b>Installer log:</b>
            <code>{INSTALL_LOG}</code>
            """,
            "success"
        )
    ))

    display(HTML(
        progress_bar(
            100,
            "Installation completed successfully"
        )
    ))

# ============================================================
# STEP 6 — ENVIRONMENT VERIFICATION
# ============================================================

print()
print("=" * 72)
print("STEP 6/6 — Verifying BindCraft environment")
print("=" * 72)

env_result = subprocess.run(
    [CONDA, "env", "list"],
    capture_output=True,
    text=True
)

if env_result.returncode != 0:
    raise RuntimeError(
        "Unable to read Conda environment list."
    )

# Locate environment
if not BINDCRAFT_ENV.exists():

    display(HTML(
        status_panel(
            "❌ BINDCRAFT ENVIRONMENT NOT FOUND",
            "Installation completed, but the BindCraft environment "
            "could not be located.",
            "error"
        )
    ))

    raise RuntimeError(
        "Installation finished but the BindCraft environment "
        "could not be located."
    )

print("✓ BindCraft environment found")
print(f"✓ Environment path: {BINDCRAFT_ENV}")

# ------------------------------------------------------------
# Environment Python
# ------------------------------------------------------------

env_python = BINDCRAFT_ENV / "bin" / "python"

if env_python.exists():

    py_result = subprocess.run(
        [str(env_python), "--version"],
        capture_output=True,
        text=True
    )

    env_python_version = (
        py_result.stdout.strip()
        or py_result.stderr.strip()
    )

    print(
        f"✓ Environment Python: "
        f"{env_python_version}"
    )

else:

    env_python_version = "Not detected"
    print("⚠️ Environment Python executable not found.")

display(HTML(progress_bar(
    100,
    "Environment installation and verification complete"
)))

# ============================================================
# FINAL STATUS
# ============================================================

display(HTML(
    status_panel(
        "🟢 FREEBINDCRAFT ENVIRONMENT READY",
        f"""
        <b>Repository:</b> ✓<br>
        <b>Conda:</b> ✓<br>
        <b>GPU:</b> ✓ A100 40 GB<br>
        <b>PyRosetta:</b> OFF<br>
        <b>BindCraft environment:</b> ✓<br>
        <b>Python:</b> {env_python_version}<br><br>

        <b>Environment path:</b><br>
        <code>{BINDCRAFT_ENV}</code>
        """,
        "success"
    )
))

print()
print("👉 USER ACTION")
print("   No additional action is required.")

print()
print("👉 NEXT STEP")
print("   Run Cell [7] to verify JAX, CUDA, OpenMM,")
print("   ColabDesign and the FreeBindCraft runtime.")

print()
print("⚠️ IMPORTANT")
print("   Do NOT start a binder-design campaign yet.")
print("   We will first perform the complete software/GPU test.")
```

```python
# ============================================================
# 🧬 FreeBindCraft — Cell 8: Upload Target PDB
# ============================================================

from google.colab import files
from pathlib import Path
from IPython.display import display, HTML, clear_output
import time
import re
import shutil

print("=" * 72)
print("🧬 FreeBindCraft — Step 8: Target Structure")
print("=" * 72)
print()
print("STATUS: 🟢 READY")
print()
print("📌 Upload the protein structure you want to design a binder against.")
print()
print("Accepted format: PDB (.pdb)")
print()

# ------------------------------------------------------------
# PROGRESS UI
# ------------------------------------------------------------

def progress_bar(percent, status):
    width = 50
    filled = int(width * percent / 100)
    empty = width - filled

    print(
        f"\r[{('█' * filled) + ('░' * empty)}] "
        f"{percent:3d}%  {status}",
        end="",
        flush=True
    )

# ------------------------------------------------------------
# UPLOAD
# ------------------------------------------------------------

print("📁 USER ACTION")
print()
print("👉 Click the upload button below and select your target PDB.")
print()

uploaded = files.upload()

print()

if not uploaded:
    print("❌ No file was uploaded.")
    raise RuntimeError("Please upload a PDB file and run this cell again.")

# ------------------------------------------------------------
# IDENTIFY PDB FILE
# ------------------------------------------------------------

pdb_files = [
    name for name in uploaded.keys()
    if name.lower().endswith(".pdb")
]

if not pdb_files:
    print("❌ No PDB file detected.")
    print("   Please upload a file with the .pdb extension.")
    raise RuntimeError("No PDB file was uploaded.")

if len(pdb_files) > 1:
    print(f"⚠️ Multiple PDB files detected: {len(pdb_files)}")
    print("   The first PDB file will be used.")

uploaded_name = pdb_files[0]

source_path = Path(uploaded_name)
target_path = Path("/content/FreeBindCraft") / uploaded_name

# Move uploaded file into project directory
shutil.move(str(source_path), str(target_path))

# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print()
print("🔄 VALIDATING TARGET STRUCTURE")
print()

progress_bar(10, "Reading uploaded PDB")
time.sleep(0.3)

if not target_path.exists():
    print("\n❌ Uploaded PDB could not be located.")
    raise RuntimeError("Target PDB file is missing.")

progress_bar(25, "Checking PDB format")
time.sleep(0.3)

text = target_path.read_text(
    errors="ignore"
)

lines = text.splitlines()

atom_lines = [
    line for line in lines
    if line.startswith(("ATOM  ", "HETATM"))
]

if not atom_lines:
    print("\n❌ No ATOM/HETATM records found.")
    raise RuntimeError(
        "The uploaded file does not appear to contain a valid PDB structure."
    )

progress_bar(45, "Detecting chains")
time.sleep(0.3)

chains = set()

for line in atom_lines:

    if len(line) >= 22:

        chain = line[21].strip()

        if chain:
            chains.add(chain)

if not chains:
    chains = {"?"}

progress_bar(65, "Counting residues")
time.sleep(0.3)

residues = set()

for line in atom_lines:

    if len(line) >= 27:

        chain = line[21].strip()
        resname = line[17:20].strip()
        resnum = line[22:26].strip()
        insertion = line[26].strip()

        residues.add(
            (chain, resnum, insertion, resname)
        )

progress_bar(80, "Checking structure records")
time.sleep(0.3)

model_records = sum(
    1 for line in lines
    if line.startswith("MODEL")
)

ter_records = sum(
    1 for line in lines
    if line.startswith("TER")
)

progress_bar(95, "Finalizing validation")
time.sleep(0.3)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

progress_bar(100, "Validation complete")
print()
print()

print("=" * 72)
print("STATUS: ✅ TARGET STRUCTURE READY")
print("=" * 72)
print()

print("📄 FILE")
print(f"   {target_path.name}")
print()

print("🧬 STRUCTURE SUMMARY")
print(f"   Chains detected : {', '.join(sorted(chains))}")
print(f"   Residues        : {len(residues)}")
print(f"   Atom records    : {len(atom_lines)}")
print(f"   MODEL records   : {model_records}")
print(f"   TER records     : {ter_records}")
print()

print("✓ PDB file found")
print("✓ ATOM/HETATM records detected")
print("✓ Chains identified")
print("✓ Residues identified")
print("✓ Basic PDB validation passed")
print()

# ------------------------------------------------------------
# STORE FOR FOLLOWING CELLS
# ------------------------------------------------------------

TARGET_PDB = str(target_path)
TARGET_CHAINS = sorted(chains)
TARGET_RESIDUE_COUNT = len(residues)
TARGET_ATOM_COUNT = len(atom_lines)

print("💾 Target information stored for the next steps.")
print()

print("👉 NEXT STEP")
print("   Run Cell [9] to select the target chain and")
print("   specify hotspot residues.")
print()
print("⚠️ Do not start a design run yet.")
```

```python
# ============================================================
# 🧬 FreeBindCraft — Cell 9: Target Definition
# ============================================================

from pathlib import Path
from IPython.display import display, HTML
import ipywidgets as widgets
import re

print("=" * 72)
print("🧬 FreeBindCraft — Step 9: Target Definition")
print("=" * 72)
print()
print("STATUS: 🟢 READY")
print()

# ============================================================
# STEP 1 — VERIFY PREVIOUS CELL
# ============================================================

if "TARGET_PDB" not in globals():
    raise RuntimeError(
        "Target PDB information is missing. "
        "Please run Cell [8] first."
    )

if "TARGET_CHAINS" not in globals():
    raise RuntimeError(
        "Target chain information is missing. "
        "Please run Cell [8] first."
    )

if not TARGET_CHAINS:
    raise RuntimeError(
        "No protein chains were detected in the uploaded PDB."
    )

# ============================================================
# TARGET SUMMARY
# ============================================================

print("📄 TARGET STRUCTURE")
print(f"   File     : {Path(TARGET_PDB).name}")
print(f"   Chains   : {', '.join(TARGET_CHAINS)}")
print(f"   Residues : {TARGET_RESIDUE_COUNT}")
print(f"   Atoms    : {TARGET_ATOM_COUNT}")
print()

print("👉 USER ACTION")
print("   1. Select the target chain.")
print("   2. Enter hotspot residues.")
print("   3. Click '🔍 Validate Target'.")
print("   4. Click 'Continue →' after successful validation.")
print()

# ============================================================
# BUILD RESIDUE MAP FROM PDB
# ============================================================

pdb_text = Path(TARGET_PDB).read_text(errors="ignore")
pdb_lines = pdb_text.splitlines()

residue_map = {}

for line in pdb_lines:

    if not line.startswith(("ATOM  ", "HETATM")):
        continue

    if len(line) < 27:
        continue

    chain = line[21].strip()
    resnum = line[22:26].strip()
    insertion = line[26].strip()

    if not chain or not resnum:
        continue

    key = (chain, resnum, insertion)
    residue_map[key] = True


# ============================================================
# COMPACT PROGRESS BAR
# ============================================================

def make_progress_html(
    percent,
    label="Progress",
    color="#28a745"
):

    percent = max(0, min(100, int(percent)))

    return f"""
    <div style="
        margin-top:12px;
        width:420px;
        max-width:100%;
        font-family:Arial,sans-serif;
    ">

        <div style="
            font-size:12px;
            font-weight:600;
            margin-bottom:5px;
            color:#e8eaed;
        ">
            {label}
        </div>

        <div style="
            width:100%;
            height:12px;
            background:#3c4043;
            border-radius:6px;
            overflow:hidden;
        ">

            <div style="
                width:{percent}%;
                height:100%;
                background:{color};
                border-radius:6px;
            ">
            </div>

        </div>

        <div style="
            font-size:11px;
            color:#b8b8b8;
            margin-top:3px;
        ">
            {percent}%
        </div>

    </div>
    """


# ============================================================
# SUCCESS RESULT PANEL
# ============================================================

def make_success_panel(
    target_chain,
    hotspots
):

    return f"""
    <div style="
        margin-top:14px;
        width:420px;
        max-width:100%;
        padding:14px 16px;
        border:1px solid #2ea043;
        border-radius:8px;
        background:#202124;
        color:#f1f3f4;
        font-family:Arial,sans-serif;
        box-sizing:border-box;
    ">

        <div style="
            color:#56d364;
            font-size:14px;
            font-weight:700;
            margin-bottom:12px;
        ">
            ✓ Target definition ready
        </div>

        <div style="
            display:grid;
            grid-template-columns:110px 1fr;
            gap:7px 10px;
            font-size:12px;
            line-height:1.5;
        ">

            <div style="color:#9aa0a6;">
                Target chain
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {target_chain}
            </div>

            <div style="color:#9aa0a6;">
                Hotspots
            </div>

            <div style="
                color:#ffffff;
                font-weight:600;
                word-break:break-word;
            ">
                {hotspots}
            </div>

        </div>

    </div>
    """


# ============================================================
# ERROR RESULT PANEL
# ============================================================

def make_error_panel(message):

    return f"""
    <div style="
        margin-top:14px;
        width:420px;
        max-width:100%;
        padding:12px 14px;
        border:1px solid #f85149;
        border-radius:8px;
        background:#202124;
        color:#f1f3f4;
        font-family:Arial,sans-serif;
        font-size:12px;
        box-sizing:border-box;
    ">

        <div style="
            color:#ff7b72;
            font-weight:700;
            margin-bottom:5px;
        ">
            ✕ Validation requires attention
        </div>

        <div style="color:#f1f3f4;">
            {message}
        </div>

    </div>
    """


# ============================================================
# WIDGETS
# ============================================================

chain_dropdown = widgets.Dropdown(
    options=TARGET_CHAINS,
    value=TARGET_CHAINS[0],
    description="Target chain:",
    style={
        "description_width": "initial"
    },
    layout=widgets.Layout(
        width="350px"
    )
)

hotspot_input = widgets.Text(
    value="",
    placeholder="Example: B56,B57,B59",
    description="Hotspots:",
    style={
        "description_width": "initial"
    },
    layout=widgets.Layout(
        width="500px"
    )
)

validate_button = widgets.Button(
    description="🔍 Validate Target",
    button_style="primary",
    layout=widgets.Layout(
        width="180px"
    )
)

continue_button = widgets.Button(
    description="Continue →",
    button_style="success",
    disabled=True,
    layout=widgets.Layout(
        width="180px"
    )
)

status_output = widgets.Output()

progress_html = widgets.HTML(
    value=make_progress_html(
        33,
        "Waiting for target definition"
    )
)

result_html = widgets.HTML(
    value=""
)


# ============================================================
# VALIDATION FUNCTION
# ============================================================

def validate_target(_):

    global TARGET_CHAIN
    global TARGET_HOTSPOTS

    with status_output:

        status_output.clear_output()

        selected_chain = chain_dropdown.value
        raw_hotspots = hotspot_input.value.strip()

        print()
        print("🔄 VALIDATING TARGET DEFINITION")
        print()

        # ----------------------------------------------------
        # CHECK INPUT
        # ----------------------------------------------------

        if not raw_hotspots:

            print("❌ Hotspot residues are required.")
            print()
            print("Example:")
            print("   B56,B57,B59")

            progress_html.value = make_progress_html(
                33,
                "Hotspot input required",
                "#f85149"
            )

            result_html.value = make_error_panel(
                "Please enter at least one hotspot residue."
            )

            continue_button.disabled = True
            return

        # ----------------------------------------------------
        # SPLIT HOTSPOTS
        # ----------------------------------------------------

        tokens = [
            x.strip()
            for x in raw_hotspots.split(",")
            if x.strip()
        ]

        print(f"✓ Entries detected: {len(tokens)}")
        print(f"✓ Selected chain: {selected_chain}")
        print()

        progress_html.value = make_progress_html(
            60,
            "Checking residue identifiers..."
        )

        # ----------------------------------------------------
        # VALIDATE RESIDUES
        # ----------------------------------------------------

        valid_hotspots = []
        invalid_hotspots = []

        for token in tokens:

            match = re.fullmatch(
                r"([A-Za-z0-9])(\d+)([A-Za-z]?)",
                token
            )

            if not match:

                invalid_hotspots.append(
                    (token, "invalid format")
                )

                continue

            chain = match.group(1)
            resnum = match.group(2)
            insertion = match.group(3)

            if chain != selected_chain:

                invalid_hotspots.append(
                    (
                        token,
                        f"belongs to chain {chain}, "
                        f"not selected chain {selected_chain}"
                    )
                )

                continue

            key = (
                chain,
                resnum,
                insertion
            )

            if key in residue_map:

                valid_hotspots.append(token)

            else:

                possible = [
                    k
                    for k in residue_map
                    if k[0] == chain
                    and k[1] == resnum
                ]

                if possible:

                    invalid_hotspots.append(
                        (
                            token,
                            "residue exists but insertion "
                            "code does not match"
                        )
                    )

                else:

                    invalid_hotspots.append(
                        (
                            token,
                            "residue not found in PDB"
                        )
                    )

        # ----------------------------------------------------
        # VALIDATION FAILURE
        # ----------------------------------------------------

        if invalid_hotspots:

            print("❌ VALIDATION FAILED")
            print()

            print("Invalid hotspot entries:")

            for token, reason in invalid_hotspots:

                print(
                    f"   • {token} — {reason}"
                )

            print()
            print(
                "Please correct the hotspot list "
                "and validate again."
            )

            progress_html.value = make_progress_html(
                66,
                "Validation failed",
                "#f85149"
            )

            result_html.value = make_error_panel(
                "One or more hotspot residues could not "
                "be validated against the selected chain."
            )

            continue_button.disabled = True
            return

        # ----------------------------------------------------
        # SUCCESS
        # ----------------------------------------------------

        TARGET_CHAIN = selected_chain

        TARGET_HOTSPOTS = ",".join(
            valid_hotspots
        )

        progress_html.value = make_progress_html(
            100,
            "Target definition complete"
        )

        print()
        print("=" * 56)
        print("STATUS: ✅ TARGET DEFINITION VALID")
        print("=" * 56)
        print()

        print(
            f"✓ Target chain : {TARGET_CHAIN}"
        )

        print(
            f"✓ Hotspots     : {TARGET_HOTSPOTS}"
        )

        print(
            f"✓ Valid residues: {len(valid_hotspots)}"
        )

        print()
        print("The target definition is ready.")

        # ----------------------------------------------------
        # NEW DARK-MODE SUCCESS PANEL
        # ----------------------------------------------------

        result_html.value = make_success_panel(
            TARGET_CHAIN,
            TARGET_HOTSPOTS
        )

        continue_button.disabled = False


# ============================================================
# CONTINUE FUNCTION
# ============================================================

def continue_to_next(_):

    with status_output:

        status_output.clear_output()

        if (
            "TARGET_CHAIN" not in globals()
            or
            "TARGET_HOTSPOTS" not in globals()
        ):

            print()
            print(
                "❌ Target definition has not been validated."
            )

            print()
            print(
                "Please click '🔍 Validate Target' first."
            )

            return

        print()
        print("=" * 56)
        print("STATUS: 🚀 TARGET DEFINITION CONFIRMED")
        print("=" * 56)
        print()

        print(
            f"✓ Target chain : {TARGET_CHAIN}"
        )

        print(
            f"✓ Hotspots     : {TARGET_HOTSPOTS}"
        )

        print()
        print("👉 NEXT STEP")
        print()
        print(
            "Run Cell [10] to configure binder design settings."
        )


# ============================================================
# CONNECT BUTTONS
# ============================================================

validate_button.on_click(
    validate_target
)

continue_button.on_click(
    continue_to_next
)


# ============================================================
# DISPLAY USER INTERFACE
# ============================================================

display(
    HTML(
        """
        <hr>

        <h3>🎯 Target Definition</h3>

        <p>
        Select the protein chain that should be targeted
        and enter the hotspot residues used to guide
        binder design.
        </p>
        """
    )
)

display(
    HTML(
        f"""
        <div style="
            margin-bottom:12px;
            font-family:Arial,sans-serif;
            font-size:13px;
        ">

        <b>📄 Target:</b>
        {Path(TARGET_PDB).name}

        <br>

        <b>🧬 Available chains:</b>
        {", ".join(TARGET_CHAINS)}

        <br>

        <b>🔢 Residues:</b>
        {TARGET_RESIDUE_COUNT}

        </div>
        """
    )
)

display(
    HTML(
        """
        <b>Target chain</b>
        """
    )
)

display(
    chain_dropdown
)

display(
    HTML(
        """
        <br>

        <b>Hotspot residues</b>

        <br>

        Enter comma-separated residue identifiers
        using the chain ID and residue number.

        <br>

        Example:
        <code>B56,B57,B59</code>

        <br><br>
        """
    )
)

display(
    hotspot_input
)

display(
    widgets.HBox(
        [
            validate_button,
            continue_button
        ]
    )
)

display(
    progress_html
)

display(
    status_output
)

display(
    result_html
)

print()
print("🟢 USER ACTION")
print()
print("1. Select the target chain.")
print("2. Enter hotspot residues.")
print("3. Click '🔍 Validate Target'.")
print("4. After successful validation, click 'Continue →'.")
```

```python
# ============================================================
# 🧬 FreeBindCraft — Cell 11: Pre-flight Validation
# ============================================================

from pathlib import Path
from IPython.display import display, HTML
import ipywidgets as widgets
import subprocess
import json
import os
import shutil
import re

print("=" * 72)
print("🧬 FreeBindCraft — Step 11: Pre-flight Validation")
print("=" * 72)
print()
print("STATUS: 🟢 READY")
print()

# ============================================================
# VERIFY PREVIOUS STEPS
# ============================================================

required_variables = [
    "TARGET_PDB",
    "TARGET_CHAIN",
    "TARGET_HOTSPOTS",
    "DESIGN_SETTINGS"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:

    raise RuntimeError(
        "Required information is missing: "
        + ", ".join(missing_variables)
        + ". Please complete the previous cells first."
    )

# ============================================================
# PATHS
# ============================================================

REPO_DIR = Path("/content/FreeBindCraft")

BINDCRAFT_SCRIPT = (
    REPO_DIR / "bindcraft.py"
)

FILTER_DIR = (
    REPO_DIR / "settings_filters"
)

ADVANCED_DIR = (
    REPO_DIR / "settings_advanced"
)

PARAMS_DIR = (
    REPO_DIR / "params"
)

FUNCTIONS_DIR = (
    REPO_DIR / "functions"
)

PROJECT_NAME = DESIGN_SETTINGS["project_name"]

OUTPUT_DIR = (
    REPO_DIR
    / "colab_results"
    / PROJECT_NAME
)

TARGET_SETTINGS_PATH = (
    OUTPUT_DIR
    / "target_settings.json"
)

FILTER_PATH = (
    FILTER_DIR
    / DESIGN_SETTINGS["filter_profile"]
)

ADVANCED_PATH = (
    ADVANCED_DIR
    / DESIGN_SETTINGS["advanced_profile"]
)

# ============================================================
# COMPACT PROGRESS BAR
# ============================================================

def make_progress_html(
    percent,
    label="Progress",
    color="#28a745"
):

    percent = max(
        0,
        min(100, int(percent))
    )

    return f"""
    <div style="
        margin-top:12px;
        width:420px;
        max-width:100%;
        font-family:Arial,sans-serif;
    ">

        <div style="
            font-size:12px;
            font-weight:600;
            margin-bottom:5px;
            color:#e8eaed;
        ">
            {label}
        </div>

        <div style="
            width:100%;
            height:12px;
            background:#3c4043;
            border-radius:6px;
            overflow:hidden;
        ">

            <div style="
                width:{percent}%;
                height:100%;
                background:{color};
                border-radius:6px;
            ">
            </div>

        </div>

        <div style="
            font-size:11px;
            color:#b8b8b8;
            margin-top:3px;
        ">
            {percent}%
        </div>

    </div>
    """


# ============================================================
# DARK SUCCESS PANEL
# ============================================================

def make_success_panel():

    return f"""
    <div style="
        margin-top:14px;
        width:520px;
        max-width:100%;
        padding:14px 16px;
        border:1px solid #2ea043;
        border-radius:8px;
        background:#202124;
        color:#f1f3f4;
        font-family:Arial,sans-serif;
        box-sizing:border-box;
    ">

        <div style="
            color:#56d364;
            font-size:14px;
            font-weight:700;
            margin-bottom:12px;
        ">
            ✓ Pre-flight validation passed
        </div>

        <div style="
            display:grid;
            grid-template-columns:145px 1fr;
            gap:7px 10px;
            font-size:12px;
            line-height:1.5;
        ">

            <div style="color:#9aa0a6;">
                Target
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {Path(TARGET_PDB).name}
            </div>

            <div style="color:#9aa0a6;">
                Target chain
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {TARGET_CHAIN}
            </div>

            <div style="color:#9aa0a6;">
                Hotspots
            </div>

            <div style="
                color:#ffffff;
                font-weight:600;
                word-break:break-word;
            ">
                {TARGET_HOTSPOTS}
            </div>

            <div style="color:#9aa0a6;">
                Binder
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {DESIGN_SETTINGS["binder_type_label"]}
            </div>

            <div style="color:#9aa0a6;">
                Length
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {DESIGN_SETTINGS["min_length"]}
                –
                {DESIGN_SETTINGS["max_length"]} aa
            </div>

            <div style="color:#9aa0a6;">
                Final designs
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {DESIGN_SETTINGS["number_of_final_designs"]}
            </div>

            <div style="color:#9aa0a6;">
                Ranking
            </div>

            <div style="
                color:#ffffff;
                font-weight:700;
            ">
                {DESIGN_SETTINGS["ranking"]}
            </div>

            <div style="color:#9aa0a6;">
                PyRosetta
            </div>

            <div style="
                color:#56d364;
                font-weight:700;
            ">
                OFF
            </div>

        </div>

    </div>
    """


# ============================================================
# DARK ERROR PANEL
# ============================================================

def make_error_panel(message):

    return f"""
    <div style="
        margin-top:14px;
        width:520px;
        max-width:100%;
        padding:12px 14px;
        border:1px solid #f85149;
        border-radius:8px;
        background:#202124;
        color:#f1f3f4;
        font-family:Arial,sans-serif;
        font-size:12px;
        box-sizing:border-box;
    ">

        <div style="
            color:#ff7b72;
            font-weight:700;
            margin-bottom:6px;
        ">
            ✕ Pre-flight validation failed
        </div>

        <div style="color:#f1f3f4;">
            {message}
        </div>

    </div>
    """


# ============================================================
# WIDGETS
# ============================================================

validate_button = widgets.Button(
    description="🔍 Run Pre-flight Check",
    button_style="primary",
    layout=widgets.Layout(
        width="200px"
    )
)

continue_button = widgets.Button(
    description="Continue →",
    button_style="success",
    disabled=True,
    layout=widgets.Layout(
        width="180px"
    )
)

status_output = widgets.Output()

progress_html = widgets.HTML(
    value=make_progress_html(
        0,
        "Waiting for pre-flight validation"
    )
)

result_html = widgets.HTML(
    value=""
)


# ============================================================
# PRE-FLIGHT VALIDATION
# ============================================================

def run_preflight(_):

    global PREFLIGHT_VALIDATION
    global TARGET_SETTINGS_PATH
    global OUTPUT_DIR

    with status_output:

        status_output.clear_output()

        continue_button.disabled = True

        print()
        print("🔄 STARTING PRE-FLIGHT VALIDATION")
        print()

        # ----------------------------------------------------
        # STEP 1 — REPOSITORY
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            10,
            "Checking FreeBindCraft repository..."
        )

        print("-" * 64)
        print("STEP 1/10 — Repository")
        print("-" * 64)

        checks = []

        if REPO_DIR.exists():

            print("✓ Repository directory found")
            checks.append(True)

        else:

            print("❌ Repository directory missing")
            checks.append(False)

        if BINDCRAFT_SCRIPT.exists():

            print("✓ bindcraft.py found")
            checks.append(True)

        else:

            print("❌ bindcraft.py missing")
            checks.append(False)

        # ----------------------------------------------------
        # STEP 2 — TARGET PDB
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            20,
            "Checking target structure..."
        )

        print()
        print("-" * 64)
        print("STEP 2/10 — Target structure")
        print("-" * 64)

        target_path = Path(TARGET_PDB)

        if target_path.exists():

            target_size = target_path.stat().st_size

            print(
                f"✓ Target PDB found"
            )

            print(
                f"✓ File size: "
                f"{target_size:,} bytes"
            )

            checks.append(True)

        else:

            print(
                f"❌ Target PDB not found:"
            )

            print(
                f"   {TARGET_PDB}"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 3 — TARGET DEFINITION
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            30,
            "Checking target definition..."
        )

        print()
        print("-" * 64)
        print("STEP 3/10 — Target definition")
        print("-" * 64)

        if TARGET_CHAIN:

            print(
                f"✓ Target chain: "
                f"{TARGET_CHAIN}"
            )

            checks.append(True)

        else:

            print("❌ Target chain missing")
            checks.append(False)

        if TARGET_HOTSPOTS:

            print(
                f"✓ Hotspots: "
                f"{TARGET_HOTSPOTS}"
            )

            checks.append(True)

        else:

            print("❌ Target hotspots missing")
            checks.append(False)

        # ----------------------------------------------------
        # STEP 4 — GPU
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            40,
            "Checking A100 GPU..."
        )

        print()
        print("-" * 64)
        print("STEP 4/10 — GPU")
        print("-" * 64)

        try:

            gpu_result = subprocess.run(
                [
                    "nvidia-smi",
                    "--query-gpu=name,memory.total,memory.free",
                    "--format=csv,noheader"
                ],
                capture_output=True,
                text=True,
                timeout=20
            )

            gpu_text = gpu_result.stdout.strip()

            print(
                f"✓ GPU detected:"
            )

            print(
                f"  {gpu_text}"
            )

            if "A100" in gpu_text:

                print(
                    "✓ A100 GPU confirmed"
                )

            else:

                print(
                    "⚠️ GPU is not an A100."
                )

            checks.append(
                gpu_result.returncode == 0
            )

        except Exception as exc:

            print(
                f"❌ GPU check failed: {exc}"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 5 — CONDA ENVIRONMENT
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            50,
            "Checking BindCraft environment..."
        )

        print()
        print("-" * 64)
        print("STEP 5/10 — Conda environment")
        print("-" * 64)

        conda_path = Path(
            "/usr/local/miniconda/envs/BindCraft/bin/python"
        )

        if conda_path.exists():

            print(
                "✓ BindCraft Python environment found"
            )

            print(
                f"  {conda_path}"
            )

            checks.append(True)

        else:

            print(
                "❌ BindCraft environment not found"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 6 — FILTER CONFIGURATION
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            60,
            "Checking filter configuration..."
        )

        print()
        print("-" * 64)
        print("STEP 6/10 — Filter configuration")
        print("-" * 64)

        if FILTER_PATH.exists():

            print(
                f"✓ Filter file found:"
            )

            print(
                f"  {FILTER_PATH.name}"
            )

            checks.append(True)

        else:

            print(
                "❌ Filter file missing:"
            )

            print(
                f"  {FILTER_PATH}"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 7 — ADVANCED CONFIGURATION
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            70,
            "Checking advanced configuration..."
        )

        print()
        print("-" * 64)
        print("STEP 7/10 — Advanced configuration")
        print("-" * 64)

        if ADVANCED_PATH.exists():

            print(
                "✓ Advanced settings found:"
            )

            print(
                f"  {ADVANCED_PATH.name}"
            )

            checks.append(True)

        else:

            print(
                "❌ Advanced settings missing:"
            )

            print(
                f"  {ADVANCED_PATH}"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 8 — ALPHAFOLD PARAMETERS
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            80,
            "Checking AlphaFold parameters..."
        )

        print()
        print("-" * 64)
        print("STEP 8/10 — AlphaFold parameters")
        print("-" * 64)

        af2_files = list(
            PARAMS_DIR.glob("params_model_*.npz")
        )

        if len(af2_files) > 0:

            print(
                f"✓ AlphaFold parameter files found: "
                f"{len(af2_files)}"
            )

            checks.append(True)

        else:

            print(
                "❌ AlphaFold parameter files not found"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 9 — TARGET SETTINGS JSON
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            90,
            "Creating target settings..."
        )

        print()
        print("-" * 64)
        print("STEP 9/10 — Target settings")
        print("-" * 64)

        OUTPUT_DIR.mkdir(
            parents=True,
            exist_ok=True
        )

        target_settings = {

            "design_path":
                str(OUTPUT_DIR),

            "binder_name":
                PROJECT_NAME,

            "starting_pdb":
                str(target_path),

            "chains":
                TARGET_CHAIN,

            "target_hotspot_residues":
                TARGET_HOTSPOTS,

            "lengths": [
                int(
                    DESIGN_SETTINGS["min_length"]
                ),
                int(
                    DESIGN_SETTINGS["max_length"]
                )
            ],

            "number_of_final_designs":
                int(
                    DESIGN_SETTINGS[
                        "number_of_final_designs"
                    ]
                )
        }

        with open(
            TARGET_SETTINGS_PATH,
            "w"
        ) as handle:

            json.dump(
                target_settings,
                handle,
                indent=2
            )

        print(
            "✓ Target settings JSON created"
        )

        print(
            f"  {TARGET_SETTINGS_PATH}"
        )

        # Verify JSON can be loaded

        try:

            with open(
                TARGET_SETTINGS_PATH
            ) as handle:

                loaded_settings = json.load(
                    handle
                )

            required_keys = [
                "design_path",
                "binder_name",
                "starting_pdb",
                "chains",
                "target_hotspot_residues",
                "lengths",
                "number_of_final_designs"
            ]

            missing_keys = [
                key
                for key in required_keys
                if key not in loaded_settings
            ]

            if missing_keys:

                print(
                    "❌ Missing target-settings keys:"
                )

                for key in missing_keys:
                    print(
                        f"   • {key}"
                    )

                checks.append(False)

            else:

                print(
                    "✓ Target settings JSON verified"
                )

                checks.append(True)

        except Exception as exc:

            print(
                f"❌ Target JSON verification failed:"
            )

            print(
                f"   {exc}"
            )

            checks.append(False)

        # ----------------------------------------------------
        # STEP 10 — CLI CHECK
        # ----------------------------------------------------

        progress_html.value = make_progress_html(
            95,
            "Checking FreeBindCraft command..."
        )

        print()
        print("-" * 64)
        print("STEP 10/10 — FreeBindCraft CLI")
        print("-" * 64)

        try:

            cli_result = subprocess.run(
                [
                    str(conda_path),
                    str(BINDCRAFT_SCRIPT),
                    "--help"
                ],
                cwd=str(REPO_DIR),
                capture_output=True,
                text=True,
                timeout=60
            )

            if cli_result.returncode == 0:

                print(
                    "✓ bindcraft.py CLI responded successfully"
                )

                cli_text = (
                    cli_result.stdout
                    + cli_result.stderr
                )

                if "--no-pyrosetta" in cli_text:

                    print(
                        "✓ --no-pyrosetta available"
                    )

                else:

                    print(
                        "⚠️ --no-pyrosetta flag not detected"
                    )

                if "--rank-by" in cli_text:

                    print(
                        "✓ --rank-by available"
                    )

                else:

                    print(
                        "⚠️ --rank-by flag not detected"
                    )

                checks.append(True)

            else:

                print(
                    "❌ bindcraft.py --help failed"
                )

                print(
                    cli_result.stderr[-1000:]
                )

                checks.append(False)

        except Exception as exc:

            print(
                f"❌ CLI check failed: {exc}"
            )

            checks.append(False)

        # ====================================================
        # FINAL RESULT
        # ====================================================

        passed = sum(
            bool(x)
            for x in checks
        )

        total = len(checks)

        print()
        print("=" * 72)

        if all(checks):

            PREFLIGHT_VALIDATION = True

            progress_html.value = make_progress_html(
                100,
                "Pre-flight validation complete"
            )

            print(
                "STATUS: ✅ PRE-FLIGHT VALIDATION PASSED"
            )

            print("=" * 72)
            print()

            print(
                f"Passed checks : "
                f"{passed}/{total}"
            )

            print(
                f"GPU           : A100"
            )

            print(
                f"Binder        : "
                f"{DESIGN_SETTINGS['binder_type_label']}"
            )

            print(
                f"Length        : "
                f"{DESIGN_SETTINGS['min_length']}"
                f"–"
                f"{DESIGN_SETTINGS['max_length']} aa"
            )

            print(
                f"Ranking       : "
                f"{DESIGN_SETTINGS['ranking']}"
            )

            print(
                f"Final designs : "
                f"{DESIGN_SETTINGS['number_of_final_designs']}"
            )

            print()
            print(
                "✓ Target settings JSON is ready."
            )

            print(
                "✓ Filter configuration is ready."
            )

            print(
                "✓ Advanced configuration is ready."
            )

            print(
                "✓ FreeBindCraft CLI is ready."
            )

            print()
            print(
                "⚠️ NO DESIGN HAS BEEN STARTED."
            )

            print(
                "The next cell will provide the final"
            )

            print(
                "RUN / CANCEL confirmation."
            )

            result_html.value = (
                make_success_panel()
            )

            continue_button.disabled = False

        else:

            PREFLIGHT_VALIDATION = False

            progress_html.value = make_progress_html(
                95,
                "Pre-flight validation failed",
                "#f85149"
            )

            print(
                "STATUS: ❌ PRE-FLIGHT VALIDATION FAILED"
            )

            print("=" * 72)
            print()

            print(
                f"Passed checks : "
                f"{passed}/{total}"
            )

            print()
            print(
                "Please correct the reported problem(s)"
            )

            print(
                "before starting a design run."
            )

            result_html.value = make_error_panel(
                f"{passed}/{total} pre-flight checks passed. "
                "Review the messages above before continuing."
            )

            continue_button.disabled = True


# ============================================================
# CONTINUE
# ============================================================

def continue_to_next(_):

    with status_output:

        status_output.clear_output()

        if not globals().get(
            "PREFLIGHT_VALIDATION",
            False
        ):

            print()
            print(
                "❌ Pre-flight validation has not passed."
            )

            print()
            print(
                "Run '🔍 Run Pre-flight Check' first."
            )

            return

        print()
        print("=" * 64)
        print(
            "STATUS: 🚀 PRE-FLIGHT CHECK CONFIRMED"
        )
        print("=" * 64)
        print()

        print(
            "✓ Target configuration ready"
        )

        print(
            "✓ Design configuration ready"
        )

        print(
            "✓ A100 GPU ready"
        )

        print(
            "✓ FreeBindCraft environment ready"
        )

        print()
        print(
            "👉 NEXT STEP"
        )

        print()
        print(
            "Run Cell [12] for the final"
        )

        print(
            "RUN / CANCEL confirmation."
        )

        print()
        print(
            "⚠️ The design run has NOT started."
        )


# ============================================================
# CONNECT BUTTONS
# ============================================================

validate_button.on_click(
    run_preflight
)

continue_button.on_click(
    continue_to_next
)


# ============================================================
# DISPLAY INTERFACE
# ============================================================

display(
    HTML(
        """
        <hr>

        <h3>🛡️ Pre-flight Validation</h3>

        <p>
        This step performs a complete safety and configuration
        check before starting a FreeBindCraft design campaign.
        </p>

        <p>
        <b>⚠️ No binder design will be started by this cell.</b>
        </p>
        """
    )
)

# Target
display(
    HTML(
        f"""
        <div style="
            margin-bottom:14px;
            padding:12px 14px;
            width:520px;
            max-width:100%;
            border:1px solid #3c4043;
            border-radius:8px;
            background:#202124;
            color:#f1f3f4;
            font-family:Arial,sans-serif;
            font-size:12px;
            box-sizing:border-box;
        ">

        <div style="
            color:#8ab4f8;
            font-weight:700;
            margin-bottom:8px;
        ">
            📄 Target configuration
        </div>

        <b>File:</b>
        {Path(TARGET_PDB).name}

        <br>

        <b>Chain:</b>
        {TARGET_CHAIN}

        <br>

        <b>Hotspots:</b>
        {TARGET_HOTSPOTS}

        <br>

        <b>Binder:</b>
        {DESIGN_SETTINGS["binder_type_label"]}

        <br>

        <b>Length:</b>
        {DESIGN_SETTINGS["min_length"]}
        –
        {DESIGN_SETTINGS["max_length"]} aa

        <br>

        <b>Final designs:</b>
        {DESIGN_SETTINGS["number_of_final_designs"]}

        <br>

        <b>Ranking:</b>
        {DESIGN_SETTINGS["ranking"]}

        </div>
        """
    )
)

# What will be checked
display(
    HTML(
        """
        <div style="
            margin-bottom:14px;
            padding:12px 14px;
            width:520px;
            max-width:100%;
            border:1px solid #3c4043;
            border-radius:8px;
            background:#202124;
            color:#f1f3f4;
            font-family:Arial,sans-serif;
            font-size:12px;
            box-sizing:border-box;
        ">

        <div style="
            color:#8ab4f8;
            font-weight:700;
            margin-bottom:8px;
        ">
            🔍 Pre-flight checks
        </div>

        ✓ Repository and bindcraft.py<br>
        ✓ Target PDB<br>
        ✓ Target chain and hotspots<br>
        ✓ A100 GPU<br>
        ✓ BindCraft Conda environment<br>
        ✓ Filter configuration<br>
        ✓ Advanced configuration<br>
        ✓ AlphaFold parameters<br>
        ✓ Target settings JSON<br>
        ✓ FreeBindCraft CLI

        </div>
        """
    )
)

# Buttons
display(
    widgets.HBox(
        [
            validate_button,
            continue_button
        ]
    )
)

# Progress
display(progress_html)

# Status
display(status_output)

# Result
display(result_html)

print()
print("🟢 USER ACTION")
print()
print("1. Click '🔍 Run Pre-flight Check'.")
print("2. Wait for all checks to complete.")
print("3. Review the validation result.")
print("4. Click 'Continue →' only if all checks pass.")
print()
print("⚠️ NO DESIGN RUN IS STARTED BY THIS CELL.")
```

```python
# ============================================================
# CELL [13] — FreeBindCraft Live Monitor
# ============================================================
# PURPOSE:
#   Monitor the running FreeBindCraft process without continuously
#   streaming output to the browser.
#
# REFRESH:
#   Every 10 minutes
#
# IMPORTANT:
#   - This monitor does NOT launch the design.
#   - It does NOT intentionally stop FreeBindCraft.
#   - FreeBindCraft continues independently between checks.
#   - The full run log remains on disk.
# ============================================================

import os
import re
import time
import json
import subprocess
from datetime import datetime, timedelta
from IPython.display import display, HTML

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

REPO_DIR = "/content/FreeBindCraft"
RUN_DIR = "/content/FreeBindCraft/colab_results/FreeBindCraft_Design"
LOG_FILE = os.path.join(RUN_DIR, "freebindcraft_run.log")

REFRESH_SECONDS = 600   # 10 minutes

# ------------------------------------------------------------
# Helper functions
# ------------------------------------------------------------

def run_cmd(cmd, timeout=20):
    try:
        result = subprocess.run(
            cmd,
            shell=True,
            capture_output=True,
            text=True,
            timeout=timeout
        )
        return result.stdout.strip()
    except Exception:
        return ""

def read_recent_log(max_chars=12000):
    """Read only the end of the log to minimise notebook/browser load."""
    if not os.path.exists(LOG_FILE):
        return ""

    try:
        with open(LOG_FILE, "rb") as f:
            f.seek(0, os.SEEK_END)
            size = f.tell()
            start = max(0, size - max_chars)
            f.seek(start)
            data = f.read()
        return data.decode("utf-8", errors="replace")
    except Exception:
        return ""

def get_gpu():
    output = run_cmd(
        "nvidia-smi "
        "--query-gpu=name,utilization.gpu,memory.used,memory.total,temperature.gpu "
        "--format=csv,noheader,nounits"
    )

    if not output:
        return {
            "name": "Unavailable",
            "util": "-",
            "used": "-",
            "total": "-",
            "temp": "-"
        }

    parts = [x.strip() for x in output.split(",")]

    if len(parts) >= 5:
        return {
            "name": parts[0],
            "util": parts[1],
            "used": parts[2],
            "total": parts[3],
            "temp": parts[4]
        }

    return {
        "name": output,
        "util": "-",
        "used": "-",
        "total": "-",
        "temp": "-"
    }

def get_process():
    output = run_cmd(
        "ps -eo pid,etime,pcpu,pmem,args | "
        "grep '[b]indcraft.py'"
    )

    if not output:
        return None

    lines = output.splitlines()

    for line in lines:
        if "--settings" in line:
            return line.strip()

    return lines[0].strip() if lines else None

def parse_status(log):
    status = {
        "trajectory": "-",
        "pLDDT": "-",
        "i_pTM": "-",
        "ipSAE": "-",
        "completed": 0,
        "failed": 0,
        "accepted": 0,
        "stage": "Waiting for run information",
        "latest": "-"
    }

    if not log:
        return status

    lines = [x.strip() for x in log.splitlines() if x.strip()]

    if lines:
        status["latest"] = lines[-1]

    # --------------------------------------------------------
    # Trajectory
    # --------------------------------------------------------

    trajectory_patterns = [
        r"Trajectory\s*[:#]?\s*([^\s,]+)",
        r"trajectory\s*[:#]?\s*([^\s,]+)",
        r"Starting trajectory\s+([^\s]+)",
        r"Running trajectory\s+([^\s]+)"
    ]

    for pattern in trajectory_patterns:
        matches = re.findall(pattern, log, flags=re.IGNORECASE)
        if matches:
            status["trajectory"] = matches[-1]

    # --------------------------------------------------------
    # pLDDT
    # --------------------------------------------------------

    matches = re.findall(
        r"pLDDT[^0-9]*([0-9]*\.?[0-9]+)",
        log,
        flags=re.IGNORECASE
    )

    if matches:
        try:
            value = float(matches[-1])
            status["pLDDT"] = f"{value:.2f}"
        except Exception:
            pass

    # --------------------------------------------------------
    # i_pTM
    # --------------------------------------------------------

    matches = re.findall(
        r"i[_\s-]?pTM[^0-9]*([0-9]*\.?[0-9]+)",
        log,
        flags=re.IGNORECASE
    )

    if matches:
        try:
            value = float(matches[-1])
            status["i_pTM"] = f"{value:.3f}"
        except Exception:
            pass

    # --------------------------------------------------------
    # ipSAE
    # --------------------------------------------------------

    matches = re.findall(
        r"ipSAE[^0-9]*([0-9]*\.?[0-9]+)",
        log,
        flags=re.IGNORECASE
    )

    if matches:
        try:
            value = float(matches[-1])
            status["ipSAE"] = f"{value:.3f}"
        except Exception:
            pass

    # --------------------------------------------------------
    # Basic event counts
    # --------------------------------------------------------

    status["failed"] = len(
        re.findall(
            r"(trajectory.*failed|trajectory.*skipped|failed.*trajectory)",
            log,
            flags=re.IGNORECASE
        )
    )

    status["completed"] = len(
        re.findall(
            r"(trajectory.*complete|trajectory.*completed|design.*complete)",
            log,
            flags=re.IGNORECASE
        )
    )

    status["accepted"] = len(
        re.findall(
            r"(accepted|final design)",
            log,
            flags=re.IGNORECASE
        )
    )

    # --------------------------------------------------------
    # Current stage
    # --------------------------------------------------------

    stage_keywords = [
        ("MPNN", "MPNN sequence design"),
        ("one-hot", "One-hot optimisation"),
        ("softmax", "Softmax optimisation"),
        ("logit", "Logit optimisation"),
        ("relax", "OpenMM relaxation"),
        ("sasa", "SASA analysis"),
        ("shape complementarity", "Shape complementarity"),
        ("analysis", "Design analysis"),
        ("AlphaFold", "AlphaFold prediction"),
        ("AF2", "AlphaFold prediction"),
        ("trajectory", "Trajectory processing")
    ]

    lower_log = log.lower()

    # Search backwards so the most recent stage wins.
    for line in reversed(lines):
        lower_line = line.lower()

        for keyword, label in stage_keywords:
            if keyword.lower() in lower_line:
                status["stage"] = label
                return status

    return status

def process_is_running():
    return get_process() is not None

def run_finished():
    """
    Detect normal completion or termination.
    """
    log = read_recent_log()

    if not log:
        return False

    completion_patterns = [
        "all trajectories completed",
        "design campaign completed",
        "campaign completed",
        "finished successfully",
        "successfully completed"
    ]

    lower = log.lower()

    return any(x in lower for x in completion_patterns)

# ------------------------------------------------------------
# Dashboard renderer
# ------------------------------------------------------------

def render_dashboard():
    now = datetime.now()

    log = read_recent_log()
    status = parse_status(log)
    gpu = get_gpu()
    process = get_process()

    running = process is not None

    if running:
        run_state = "RUNNING"
        state_symbol = "🟢"
    elif run_finished():
        run_state = "COMPLETED"
        state_symbol = "✅"
    else:
        run_state = "NOT RUNNING"
        state_symbol = "⚪"

    next_check = now + timedelta(seconds=REFRESH_SECONDS)

    # --------------------------------------------------------
    # HTML
    # --------------------------------------------------------

    html = f"""
    <div style="
        font-family:Arial,sans-serif;
        background:#111827;
        color:#f9fafb;
        padding:18px;
        border-radius:12px;
        max-width:760px;
        border:1px solid #374151;
        box-sizing:border-box;
    ">

      <div style="
          font-size:20px;
          font-weight:700;
          margin-bottom:6px;
      ">
        🧬 FreeBindCraft Live Monitor
      </div>

      <div style="
          font-size:14px;
          color:#d1d5db;
          margin-bottom:14px;
      ">
        {state_symbol} Design process: <b>{run_state}</b>
      </div>

      <!-- IMPORTANT USER NOTICE -->
      <div style="
          background:#1f2937;
          border:1px solid #4b5563;
          border-radius:9px;
          padding:12px 14px;
          margin-bottom:16px;
      ">
        <div style="
            font-size:15px;
            font-weight:700;
            margin-bottom:5px;
        ">
          ⏱️ Monitor refresh interval: <b>10 minutes</b>
        </div>

        <div style="
            font-size:13px;
            color:#d1d5db;
            line-height:1.5;
        ">
          This monitor checks the FreeBindCraft run once every
          <b>10 minutes</b>. The design process continues running
          independently between checks.
        </div>

        <div style="
            font-size:12px;
            color:#9ca3af;
            margin-top:7px;
        ">
          Last checked: {now.strftime("%Y-%m-%d %H:%M:%S")}
          &nbsp;•&nbsp;
          Next check: approximately {next_check.strftime("%H:%M:%S")}
        </div>
      </div>

      <!-- STATUS GRID -->
      <div style="
          display:grid;
          grid-template-columns:repeat(2,minmax(0,1fr));
          gap:8px;
          margin-bottom:14px;
      ">

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">CURRENT STAGE</div>
          <div style="font-size:14px;font-weight:600;margin-top:3px;">
            {status["stage"]}
          </div>
        </div>

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">TRAJECTORY</div>
          <div style="font-size:14px;font-weight:600;margin-top:3px;">
            {status["trajectory"]}
          </div>
        </div>

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">pLDDT</div>
          <div style="font-size:17px;font-weight:700;margin-top:3px;">
            {status["pLDDT"]}
          </div>
        </div>

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">i_pTM</div>
          <div style="font-size:17px;font-weight:700;margin-top:3px;">
            {status["i_pTM"]}
          </div>
        </div>

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">ipSAE</div>
          <div style="font-size:17px;font-weight:700;margin-top:3px;">
            {status["ipSAE"]}
          </div>
        </div>

        <div style="background:#1f2937;padding:10px;border-radius:8px;">
          <div style="font-size:11px;color:#9ca3af;">COMPLETED / FAILED</div>
          <div style="font-size:17px;font-weight:700;margin-top:3px;">
            {status["completed"]} / {status["failed"]}
          </div>
        </div>

      </div>

      <!-- GPU -->
      <div style="
          background:#1f2937;
          padding:12px;
          border-radius:9px;
          margin-bottom:12px;
      ">

        <div style="
            font-size:12px;
            color:#9ca3af;
            margin-bottom:6px;
        ">
          GPU STATUS
        </div>

        <div style="font-size:14px;font-weight:600;">
          🖥️ {gpu["name"]}
        </div>

        <div style="
            display:grid;
            grid-template-columns:repeat(3,1fr);
            gap:8px;
            margin-top:8px;
            font-size:12px;
        ">

          <div>
            <span style="color:#9ca3af;">Utilisation</span><br>
            <b>{gpu["util"]}%</b>
          </div>

          <div>
            <span style="color:#9ca3af;">VRAM</span><br>
            <b>{gpu["used"]} / {gpu["total"]} MiB</b>
          </div>

          <div>
            <span style="color:#9ca3af;">Temperature</span><br>
            <b>{gpu["temp"]}°C</b>
          </div>

        </div>
      </div>

      <!-- PROCESS -->
      <div style="
          background:#1f2937;
          padding:10px;
          border-radius:8px;
          margin-bottom:12px;
      ">
        <div style="font-size:11px;color:#9ca3af;">
          FREEBINDCRAFT PROCESS
        </div>

        <div style="
            font-size:12px;
            margin-top:4px;
            word-break:break-all;
        ">
          {"Running independently in background." if running else "No active bindcraft.py process detected."}
        </div>
      </div>

      <!-- LOG -->
      <div style="
          background:#0b1220;
          padding:10px;
          border-radius:8px;
          border:1px solid #374151;
      ">
        <div style="font-size:11px;color:#9ca3af;margin-bottom:5px;">
          LATEST LOG EVENT
        </div>

        <div style="
            font-family:monospace;
            font-size:11px;
            line-height:1.4;
            word-break:break-word;
            color:#d1d5db;
        ">
          {status["latest"][:500]}
        </div>
      </div>

      <div style="
          font-size:11px;
          color:#6b7280;
          margin-top:10px;
      ">
        Full log:
        {LOG_FILE}
      </div>

    </div>
    """

    return HTML(html)

# ------------------------------------------------------------
# Initial message
# ------------------------------------------------------------

print("=" * 72)
print("🧬 FreeBindCraft Live Monitor")
print("=" * 72)
print()
print("⏱️  Refresh interval: EVERY 10 MINUTES")
print()
print("The monitor performs a lightweight status check every 10 minutes.")
print("FreeBindCraft continues running independently between checks.")
print()
print("🚫 This cell does NOT launch the design.")
print("🚫 This cell does NOT intentionally terminate FreeBindCraft.")
print()
print(f"📄 Log file:")
print(f"   {LOG_FILE}")
print()
print("Starting monitor...")
print("=" * 72)

# ------------------------------------------------------------
# Display/update loop
# ------------------------------------------------------------

display_id = None

while True:

    dashboard = render_dashboard()

    if display_id is None:
        display_id = display(dashboard, display_id=True)
    else:
        display_id.update(dashboard)

    # Check whether the design process is still alive.
    if not process_is_running():

        print()
        print("=" * 72)

        if run_finished():
            print("✅ FreeBindCraft appears to have completed.")
        else:
            print("ℹ️ FreeBindCraft process is no longer detected.")

        print("=" * 72)
        print()
        print("The live monitor has stopped automatically.")
        print(f"Full run log: {LOG_FILE}")

        break

    # --------------------------------------------------------
    # Sleep for 10 minutes.
    #
    # During this time the FreeBindCraft process continues
    # independently.
    # --------------------------------------------------------

    time.sleep(REFRESH_SECONDS)
```
